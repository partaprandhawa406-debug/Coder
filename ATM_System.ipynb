```python
def main():
    balance = 25000
    correct_pin = "1234"

    print("======================================")
    print("             ATM SYSTEM")
    print("======================================")

    pin = input("Enter PIN: ")

    if pin == correct_pin:

        print("\nPIN Correct")
        print("Welcome to ATM")

        # ACCOUNT TYPE
        print("\n--------------------------------------")
        print("           ACCOUNT TYPE")
        print("--------------------------------------")
        print("1. Saving Account")
        print("2. Current Account")

        account = input("Enter Account Type: ")

        if account == "1":
            account_type = "Saving Account"

        elif account == "2":
            account_type = "Current Account"

        else:
            print("Invalid Account Type")
            account_type = "Saving Account"

        print("\nSelected Account:", account_type)

        # ATM MENU
        print("\n--------------------------------------")
        print("              ATM MENU")
        print("--------------------------------------")

        print("1. Balance")
        print("2. Withdraw")
        print("3. Deposit")
        print("4. Check Account")
        print("5. Change PIN")
        print("6. UPI Transfer")
        print("7. UPI Withdraw")
        print("8. Exit")

        operation = input("\nEnter operation: ")

        # BALANCE
        if operation == "Balance" or operation == "1":

            print("\n----- BALANCE -----")
            print("Account Type:", account_type)
            print("Your Balance is:", balance)

        # WITHDRAW
        elif operation == "Withdraw" or operation == "2":

            print("\n----- WITHDRAW -----")
            print("Maximum withdrawal limit: 10000")

            amount = int(input("Enter amount: "))

            if amount > 10000:
                print("Maximum withdrawal limit is 10000")

            elif amount <= 0:
                print("Invalid Amount")

            elif amount > balance:
                print("Insufficient Balance")

            elif amount % 100 != 0:
                print("Please enter amount in multiples of 100")

            else:
                balance = balance - amount

                print("\nWithdrawal Successful")
                print("Withdraw Amount:", amount)
                print("New Balance:", balance)

        # DEPOSIT
        elif operation == "Deposit" or operation == "3":

            print("\n----- DEPOSIT -----")
            print("Maximum deposit limit: 50000")

            amount = int(input("Enter amount: "))

            if amount <= 0:
                print("Invalid Amount")

            elif amount > 50000:
                print("Maximum deposit limit is 50000")

            else:
                balance = balance + amount

                print("\nDeposit Successful")
                print("Deposit Amount:", amount)
                print("New Balance:", balance)

        # CHECK ACCOUNT
        elif operation == "Check Account" or operation == "4":

            print("\n----- ACCOUNT DETAILS -----")

            print("Account Type:", account_type)
            print("Account Status: Active")
            print("Current Balance:", balance)

            if balance >= 20000:
                print("Account Balance: Excellent")

            elif balance >= 10000:
                print("Account Balance: Good")

            elif balance > 0:
                print("Account Balance: Low")

            else:
                print("Account Balance: Zero")

        # CHANGE PIN
        elif operation == "Change PIN" or operation == "5":

            print("\n----- CHANGE PIN -----")

            old_pin = input("Enter Old PIN: ")

            if old_pin == correct_pin:

                new_pin = input("Enter New PIN: ")

                if len(new_pin) == 4:

                    correct_pin = new_pin

                    print("\nPIN Changed Successfully")
                    print("Your New PIN is:", correct_pin)

                else:
                    print("PIN must be 4 digits")

            else:
                print("Wrong Old PIN")

        # UPI TRANSFER
        elif operation == "UPI Transfer" or operation == "6":

            print("\n----- UPI TRANSFER -----")

            upi_id = input("Enter UPI ID: ")
            amount = int(input("Enter Amount: "))

            if amount <= 0:

                print("Invalid Amount")

            elif amount > 50000:

                print("UPI Transfer Limit is 10000")

            elif amount > balance:

                print("Insufficient Balance")

            else:

                balance = balance - amount

                print("\nUPI Transfer Successful")
                print("UPI ID:", upi_id)
                print("Transfer Amount:", amount)
                print("Remaining Balance:", balance)


        elif operation == "UPI Withdraw" or operation == "7":

            print("\n----- UPI WITHDRAW -----")

            upi_id = input("Enter UPI ID: ")
            amount = int(input("Enter Amount: "))

            if amount <= 0:

                print("Invalid Amount")

            elif amount > 10000:

                print("UPI Withdraw Limit is 10000")

            elif amount > balance:

                print("Insufficient Balance")

            else:

                balance = balance - amount

                print("\nUPI Withdrawal Successful")
                print("UPI ID:", upi_id)
                print("Withdraw Amount:", amount)
                print("Remaining Balance:", balance)

        elif operation == "Exit" or operation == "8":
            print("\nThank You")
            print("Please Collect Your Card")
        else:
            print("\nInvalid Operation")
            print("Please select a valid option")

    else:
        print("\n================================")
        print("          ACCESS DENIED")
        print("================================")
        print("Wrong PIN")
        print("Please try again with correct PIN")

    print("\n======================================")
    print("        THANK YOU FOR USING ATM")
    print("======================================")

if __name__ == '__main__':
    main()
```
