```python
import pandas as pd 
```

```python
df = pd.read_csv("Nike_Sales_Uncleaned.csv")
```

```python
df.head()
```

```python
size_map = {
    "Kids": {"S": 3, "M": 4, "L": 6, "XL": 7},
    "Women": {"S": 5, "M": 7, "L": 8, "XL": 10},
    "Men": {"S": 6, "M": 8, "L": 10, "XL": 12}
}

df["Size"] = df.apply(
    lambda r: size_map.get(r["Gender_Category"], {}).get(
        str(r["Size"]).strip(),
        pd.to_numeric(r["Size"], errors="coerce")
    ),
    axis=1
)
```

```python
df.head(50)
```

```python
df["Size"].isnull().sum()
```

```python
df[df["Size"].isnull()]["Gender_Category"].value_counts()
```

```python
df.groupby("Gender_Category")["Size"].median()
```

```python
df["Size"] = df["Size"].fillna(
    df.groupby("Gender_Category")["Size"].transform("median")
)
```

```python
df["Size"].isnull().sum()
```
