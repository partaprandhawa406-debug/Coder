```python
def main():
    print("==============================================")
    print("          MOVIE TICKET BOOKING SYSTEM")
    print("==============================================")

    # Customer Details
    name = input("Enter Customer Name: ")
    movie = input("Enter Movie Name: ")
    age = int(input("Enter Your Age: "))
    tickets = int(input("Enter Number of Tickets: "))

    print("\n----------------------------------------------")
    print("              MOVIE DETAILS")
    print("----------------------------------------------")

    print("Customer Name :", name)
    print("Movie Name    :", movie)
    print("Age           :", age)
    print("Tickets       :", tickets)


    # Age Check
    if age >= 18:

        print("\n----------------------------------------------")
        print("              TICKET OPTIONS")
        print("----------------------------------------------")

        print("1. Normal Seat - ₹200")
        print("2. Premium Seat - ₹300")
        print("3. VIP Seat - ₹500")

        seat = int(input("Select Seat Type (1-3): "))

        # Seat Price
        if seat == 1:
            seat_name = "Normal"
            price = 200

        elif seat == 2:
            seat_name = "Premium"
            price = 300

        elif seat == 3:
            seat_name = "VIP"
            price = 500

        else:
            seat_name = "Normal"
            price = 200
            print("Invalid choice! Normal seat selected.")


        # Ticket Amount
        total = tickets * price


        # Discount
        print("\n----------------------------------------------")
        print("              DISCOUNT CHECK")
        print("----------------------------------------------")

        if tickets >= 5:
            discount = total * 10 / 100

        elif tickets >= 3:
            discount = total * 5 / 100

        else:
            discount = 0


        # Final Amount
        final_amount = total - discount


        # Booking Details
        print("\n==============================================")
        print("             BOOKING SUCCESSFUL")
        print("==============================================")

        print("Customer Name  :", name)
        print("Movie Name     :", movie)
        print("Age            :", age)
        print("Seat Type      :", seat_name)
        print("Number Tickets :", tickets)

        print("----------------------------------------------")

        print("Ticket Price   : ₹", price)
        print("Total Amount   : ₹", total)
        print("Discount       : ₹", discount)
        print("Final Amount   : ₹", final_amount)

        print("----------------------------------------------")
        print("Your movie tickets have been booked!")
        print("Please reach the theatre on time.")


    else:

        print("\n==============================================")
        print("             BOOKING NOT ALLOWED")
        print("==============================================")

        print("Customer Name :", name)
        print("Movie Name    :", movie)
        print("Age           :", age)

        print("----------------------------------------------")
        print("You must be 18 or older to book this movie.")
        print("Booking cannot be completed.")


    print("\n==============================================")
    print("       THANK YOU FOR USING OUR SYSTEM")
    print("==============================================")

if __name__ == '__main__':
    main()
```
