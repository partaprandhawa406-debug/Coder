```python
def main():
    print("=================================")
    print("===== HOSPITAL MANAGEMENT =====")
    print("==*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*d")

    id = input("Patient ID: ")
    name = input("Name: ")
    age = int(input("Age: "))
    disease = input("Disease: ")

    print("\n1.General  2.Cardiology  3.Orthopedic")
    d = int(input("Department: "))

    if d == 1:
        dept, cf = "General", 500
    elif d == 2:
        dept, cf = "Cardiology", 1200
    else:
        dept, cf = "Orthopedic", 900

    print("\n1.General Doctor  2.Specialist  3.Senior Specialist")
    x = int(input("Doctor: "))

    if x == 1:
        doctor, df = "General Doctor", 500
    elif x == 2:
        doctor, df = "Specialist", 1000
    else:
        doctor, df = "Senior Specialist", 1500

    print("\n1.Blood Test  2.X-Ray  3.MRI")
    t = int(input("Test: "))

    if t == 1:
        test, tf = "Blood Test", 500
    elif t == 2:
        test, tf = "X-Ray", 800
    else:
        test, tf = "MRI", 2500

    print("\n1.General Ward  2.Private  3.ICU")
    r = int(input("Room: "))
    days = int(input("Days: "))

    if r == 1:
        room, rf = "General Ward", 1000
    elif r == 2:
        room, rf = "Private", 4000
    else:
        room, rf = "ICU", 7000

    bill = cf + df + tf + (rf * days)

    if age < 5:
        discount = 20
    elif age < 18:
        discount = 10
    else:
        discount = 0

    final_bill = bill - (bill * discount / 100)

    print("\n===== PATIENT REPORT =====")
    print("ID:", id)
    print("Name:", name)
    print("Age:", age)
    print("Disease:", disease)
    print("Department:", dept)
    print("Doctor:", doctor)
    print("Test:", test)
    print("Room:", room)
    print("Days:", days)
    print("Total Bill:", bill)
    print("Discount:", discount, "%")
    print("Final Bill:", final_bill)
    print("==========================")

if __name__ == '__main__':
    main()
```
