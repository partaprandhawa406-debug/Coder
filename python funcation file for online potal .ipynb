```python
def apply_twice(func, value):
    return func(func(value))

def double(number):
    return number * 2

print(apply_twice(double, 5))
```

```python
def apply_to_list(numbers, func):
    result = []
    for number in numbers:
        result.append(func(number))
    return result

numbers = [1, 2, 3, 4, 5]
print(apply_to_list(numbers, lambda x: x * x))
```

```python
def power_function(power):
    def calculate_power(number):
        return number ** power
    return calculate_power

square = power_function(2)
cube = power_function(3)

print(square(5))
print(cube(3))
```

```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

def calculator(a, b, selected_function):
    return selected_function(a, b)

print(calculator(10, 5, addition))
print(calculator(10, 5, subtraction))
print(calculator(10, 5, multiplication))
print(calculator(10, 5, division))
```

```python
def check_grade(marks):
    if marks >= 40:
        return "Pass"
    return "Fail"

def student_result(marks, grading_function):
    return grading_function(marks)

print(student_result(75, check_grade))
print(student_result(30, check_grade))
```

```python
def convert_temperature(temperatures, conversion_function):
    return [conversion_function(temp) for temp in temperatures]

def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

temperatures = [0, 20, 30, 40]
print(convert_temperature(temperatures, celsius_to_fahrenheit))
```

```python
def discount(price):
    return price * 0.10

def final_bill(price, discount_function):
    discount_amount = discount_function(price)
    return price - discount_amount

print(final_bill(1000, discount))
```

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

operations = {
    "+": add,
    "-": subtract,
    "*": multiply,
    "/": divide
}

num1 = 20
num2 = 5
operator = "*"

if operator in operations:
    print("Result:", operations[operator](num1, num2))
else:
    print("Invalid operator")
```
