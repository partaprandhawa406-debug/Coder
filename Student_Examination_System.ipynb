```python
def main():

    print("========================================")
    print("   STUDENT EXAMINATION SYSTEM")
    print("========================================")

    username = input("Enter Admin Username: ")
    password = input("Enter Admin Password: ")

    if username == "admin" and password == "1234":

        print("\nAdmin Login Successful!")

        print("\n----------------------------------------")
        print("          STUDENT DETAILS")
        print("----------------------------------------")

        name = input("Enter Student Name: ")
        cls = input("Enter Class: ")
        roll = input("Enter Roll Number: ")

        print("\nEnter Marks Between 0 and 80")

        s1 = int(input("English Marks: "))
        s2 = int(input("Maths Marks: "))
        s3 = int(input("Science Marks: "))
        s4 = int(input("Computer Marks: "))
        s5 = int(input("Punjabi Marks: "))

        if s1 < 0 or s1 > 80 or s2 < 0 or s2 > 80 or s3 < 0 or s3 > 80 or s4 < 0 or s4 > 80 or s5 < 0 or s5 > 80:

            print("\nERROR: Invalid Marks!")
            print("Every Subject Must Have Marks From 0 to 80.")
            print("Report Card Cannot Be Generated.")

        else:

            total = s1 + s2 + s3 + s4 + s5
            percentage = total / 400 * 100

            if s1 >= 27 and s2 >= 27 and s3 >= 27 and s4 >= 27 and s5 >= 27:
                result = "PASS"
            else:
                result = "FAIL"

            if percentage >= 80:
                grade = "A+"
            else:
                if percentage >= 70:
                    grade = "A"
                else:
                    if percentage >= 60:
                        grade = "B"
                    else:
                        if percentage >= 50:
                            grade = "C"
                        else:
                            if percentage >= 40:
                                grade = "D"
                            else:
                                grade = "F"

            print("\n========================================")
            print("              REPORT CARD")
            print("========================================")
            print("Student Name :", name)
            print("Class        :", cls)
            print("Roll Number  :", roll)
            print("----------------------------------------")
            print("English      :", s1, "/ 80")
            print("Maths        :", s2, "/ 80")
            print("Science      :", s3, "/ 80")
            print("Computer     :", s4, "/ 80")
            print("Punjabi      :", s5, "/ 80")
            print("----------------------------------------")
            print("Total Marks  :", total, "/ 400")
            print("Percentage   :", round(percentage, 2), "%")
            print("Grade        :", grade)
            print("Final Result :", result)
            print("========================================")

            if result == "PASS":
                print("Congratulations! Student Passed.")
            else:
                print("Student Failed. Please Work Hard.")

    else:
        print("\nERROR: Wrong Admin Username or Password!")
        print("Access Denied.")

if __name__ == '__main__':
    main()
```
