```python
def main():
    # ========================================
    #         LOAN ELIGIBILITY CHECKER
    # ========================================

    # 1. Take User Inputs
    name = input("Enter Applicant Name: ")
    age = int(input("Enter Age: "))
    salary = float(input("Enter Monthly Salary: "))
    employment = input("Enter Employment Type (Salaried / Self-Employed / Unemployed): ")
    experience = float(input("Enter Experience (in years): "))
    credit = int(input("Enter Credit Score: "))
    emi = float(input("Enter Existing Monthly EMI: "))
    loan_amount = float(input("Enter Required Loan Amount: "))

    remaining_salary = salary - emi

    if age >= 21 and age <= 60:
        if employment != "partap":
            if salary >= 15000:
                if credit >= 650:
                    if experience >= 1:
                        if remaining_salary >= 15000:
                            if salary >= 50000 and loan_amount <= 1000000:
                                status = "ELIGIBLE"
                                reason = "All conditions met successfully"
                            elif salary < 50000 and loan_amount <= 500000:
                                status = "ELIGIBLE"
                                reason = "All conditions met successfully"
                            else:
                                status = "NOT ELIGIBLE"
                                reason = "Loan amount requested is too high for your salary"
                        else:
                            status = "NOT ELIGIBLE"
                            reason = "High Financial Risk (Remaining salary after EMI is below 15000)"
                    else:
                        status = "NOT ELIGIBLE"
                        reason = "Work Experience is Too Low"
                else:
                    status = "NOT ELIGIBLE"
                    reason = "Credit Score Too Low"
            else:
                status = "NOT ELIGIBLE"
                reason = "Monthly Salary Below Minimum Limit (15000)"
        else:
            status = "NOT ELIGIBLE"
            reason = "Applicant is Unemployed"
    else:
        status = "NOT ELIGIBLE"
        reason = "Age Must Be Between 21 and 60"


    # 3. Final Report Output
    print("\n========================================")
    print("         LOAN ELIGIBILITY REPORT        ")
    print("========================================")
    print("Applicant Name      :", name)
    print("Age                 :", age)
    print("Monthly Salary      :", salary)
    print("Credit Score        :", credit)
    print("----------------------------------------")
    print("Loan Status         :", status)
    print("Reason              :", reason)
    print("========================================")

if __name__ == '__main__':
    main()
```
