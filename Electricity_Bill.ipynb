```python
def main():
    print("==============================================")
    print("          ELECTRICITY BILL GENERATOR")
    print("==============================================")

    # Customer Details

    name = input("Enter Customer Name: ")
    father_name = input("Enter Father's Name: ")
    meter_no = input("Enter Meter Number: ")
    address = input("Enter Address: ")
    phone = input("Enter Phone Number: ")

    # Electricity Details

    units = int(input("Enter Electricity Units: "))

    print("\n----------------------------------------------")
    print("             CUSTOMER DETAILS")
    print("----------------------------------------------")

    print("Customer Name :", name)
    print("Father Name   :", father_name)
    print("Meter Number  :", meter_no)
    print("Address       :", address)
    print("Phone Number  :", phone)
    print("Units Used    :", units)

    # Bill Calculation

    if units <= 100:
        rate = 5
        bill = units * rate

    elif units <= 200:
        rate = 7
        bill = units * rate

    else:
        rate = 10
        bill = units * rate

    # Discount

    if bill >= 2000:
        discount = 200
    else:
        discount = 0

    final_bill = bill - discount

    # Final Bill

    print("\n----------------------------------------------")
    print("             ELECTRICITY BILL")
    print("----------------------------------------------")

    print("Customer Name :", name)
    print("Meter Number  :", meter_no)
    print("Units Used    :", units)
    print("Rate Per Unit :", rate)
    print("Bill Amount   : Rs.", bill)
    print("Discount      : Rs.", discount)
    print("Final Bill    : Rs.", final_bill)

    print("----------------------------------------------")

    if final_bill > 0:
        print("Payment Status: Bill Generated Successfully")
    else:
        print("Payment Status: No Bill")

    print("----------------------------------------------")
    print("        THANK YOU FOR USING OUR SERVICE")
    print("==============================================")

if __name__ == '__main__':
    main()
```
