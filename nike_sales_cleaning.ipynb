```
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', dayfirst=True, errors='coerce')
```

```
for col in ['Gender_Category', 'Product_Line', 'Product_Name', 'Sales_Channel', 'Region']:
    df[col] = df[col].astype('string').str.strip()
```

```
df['Gender_Category'] = df['Gender_Category'].str.title()
df['Product_Line'] = df['Product_Line'].str.title()
df['Sales_Channel'] = df['Sales_Channel'].str.title()
df['Region'] = df['Region'].str.lower().replace({
    'bengaluru': 'Bangalore',
    'bangalore': 'Bangalore',
    'hyd': 'Hyderabad',
    'hyderbad': 'Hyderabad',
    'hyderabad': 'Hyderabad'
}).str.title()
```

```
df['Size'] = df['Size'].astype('string').str.strip().str.upper()
df['Size'] = df['Size'].fillna('Unknown')
```

```
df['Units_Sold'] = pd.to_numeric(df['Units_Sold'], errors='coerce')
df.loc[df['Units_Sold'] < 0, 'Units_Sold'] = np.nan
df['Units_Sold'] = df['Units_Sold'].fillna(df['Units_Sold'].median())
```

```
df['MRP'] = pd.to_numeric(df['MRP'], errors='coerce')
df.loc[df['MRP'] <= 0, 'MRP'] = np.nan
df['MRP'] = df['MRP'].fillna(df['MRP'].median())
```

```
df['Discount_Applied'] = pd.to_numeric(df['Discount_Applied'], errors='coerce')
df.loc[~df['Discount_Applied'].between(0, 1), 'Discount_Applied'] = np.nan
df['Discount_Applied'] = df['Discount_Applied'].fillna(df['Discount_Applied'].median())
```

```
df['Revenue'] = pd.to_numeric(df['Revenue'], errors='coerce')
df.loc[df['Revenue'] < 0, 'Revenue'] = np.nan
calculated_revenue = df['Units_Sold'] * df['MRP'] * (1 - df['Discount_Applied'])
df['Revenue'] = df['Revenue'].fillna(calculated_revenue)
```

```
df['Revenue'] = df['Revenue'].fillna(df['Revenue'].median())
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
```

```
df = df.drop_duplicates().reset_index(drop=True)
```
