```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
sns.countplot(data=tips, x="day")
plt.title("Number of Records by Day")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", palette="Set1")
plt.title("Total Bill and Tip by Gender")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
students = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Subject": ["Python", "Python", "Math", "Math", "English", "English"],
    "Marks": [85, 90, 65, 75, 80, 95],
    "Result": ["Pass", "Pass", "Pass", "Pass", "Pass", "Pass"]
})

# Count plot
sns.countplot(data=students, x="Subject", hue="Gender")
plt.title("Students by Subject and Gender")
plt.show()

# Bar plot
sns.barplot(data=students, x="Subject", y="Marks")
plt.title("Average Marks by Subject")
plt.show()

# Heatmap
sns.heatmap(students[["Marks"]].corr(), annot=True, cmap="coolwarm")
plt.title("Student Marks Correlation")
plt.show()
```
