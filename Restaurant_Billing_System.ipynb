```python
def main():
    print("==========================================")
    print("          RESTAURANT BILLING SYSTEM")
    print("==========================================")

    # Customer Details
    name = input("Enter Customer Name: ")
    table = int(input("Enter Table Number: "))

    print("\n----------- MENU -----------")

    print("1. Pizza      - Rs. 150")
    print("2. Burger     - Rs. 80")
    print("3. Momos      - Rs. 70")
    print("4. Cold Drink - Rs. 40")

    print("----------------------------")

    choice = int(input("Enter your choice: "))
    quantity = int(input("Enter quantity: "))


    # Item Selection
    if choice == 1:
        item = "Pizza"
        price = 150

    elif choice == 2:
        item = "Burger"
        price = 80

    elif choice == 3:
        item = "Momos"
        price = 70

    elif choice == 4:
        item = "Cold Drink"
        price = 40

    else:
        item = "Invalid"
        price = 0


    # Calculate Total
    total = price * quantity


    # Check Discount
    if total >= 500:
        discount = total * 10 / 100

    elif total >= 300:
        discount = total * 5 / 100

    else:
        discount = 0


    # Amount After Discount
    amount_after_discount = total - discount


    # GST Calculation
    gst = amount_after_discount * 5 / 100


    # Final Bill
    final_amount = amount_after_discount + gst


    # Print Bill
    print("\n==========================================")
    print("              RESTAURANT BILL")
    print("==========================================")

    print("Customer Name :", name)
    print("Table Number  :", table)

    print("------------------------------------------")

    print("Item          :", item)
    print("Quantity      :", quantity)
    print("Price         : Rs.", price)

    print("------------------------------------------")

    print("Total Amount  : Rs.", total)
    print("Discount      : Rs.", discount)
    print("GST (5%)      : Rs.", gst)

    print("------------------------------------------")

    print("Final Amount  : Rs.", final_amount)

    print("==========================================")
    print("        THANK YOU FOR VISITING!")
    print("==========================================")

if __name__ == '__main__':
    main()
```
