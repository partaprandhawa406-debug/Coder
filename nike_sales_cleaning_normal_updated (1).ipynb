```python
df = df.drop_duplicates()
```

```python

for col in ['Gender_Category', 'Product_Line', 'Product_Name', 'Size', 'Sales_Channel', 'Region']:
    df[col] = df[col].astype('string').str.strip()
```

```python
df['Region'] = df['Region'].replace({'bengaluru': 'Bangalore', 'Bengaluru': 'Bangalore', 'Hyd': 'Hyderabad'})
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', errors='coerce')
```

```python
df['Units_Sold'] = df['Units_Sold'].fillna(df['Units_Sold'].median())
df['MRP'] = df['MRP'].fillna(df['MRP'].median())
df['Discount_Applied'] = df['Discount_Applied'].fillna(df['Discount_Applied'].median())
```

```python
df['Size'] = df['Size'].fillna('Unknown')
df['Order_Date'] = df['Order_Date'].fillna(df['Order_Date'].mode()[0])
```

```python
for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].fillna('Not')
```
