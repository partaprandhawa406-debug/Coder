```python
def main():

    print("==============================================")
    print("          SCHOOL ADMISSION SYSTEM")
    print("==============================================")

    # Student Information

    name = input("Enter Student Name: ")
    father = input("Enter Father's Name: ")
    mother = input("Enter Mother's Name: ")

    age = int(input("Enter Student Age: "))
    student_class = int(input("Enter Student Class: "))

    gender = input("Enter Gender: ")
    phone = input("Enter Phone Number: ")
    address = input("Enter Address: ")

    fee = int(input("Enter Admission Fee: "))

    print("\n==============================================")
    print("             STUDENT DETAILS")
    print("==============================================")

    print("Student Name :", name)
    print("Father Name  :", father)
    print("Mother Name  :", mother)
    print("Age          :", age)
    print("Class        :", student_class)
    print("Gender       :", gender)
    print("Phone Number :", phone)
    print("Address      :", address)
    print("Fee          :", fee)

    print("\n==============================================")
    print("           ADMISSION CHECK")
    print("==============================================")

    # Main condition
    if age >= 5:

        print("Age Check: Passed")

        # Nested class condition
        if student_class >= 1 and student_class <= 5:

            print("Class Check: Primary School")

            # Nested fee condition
            if fee >= 5000:

                print("Fee Check: Passed")
                print("\nAdmission Status: SUCCESSFUL")
                print("Welcome to Primary School")

            elif fee >= 3000:

                print("Fee Check: Partially Passed")
                print("Please pay remaining fee")
                print("Admission Status: PENDING")

            else:

                print("Fee Check: Failed")
                print("Admission Status: NOT SUCCESSFUL")

        elif student_class >= 6 and student_class <= 8:

            print("Class Check: Middle School")

            if fee >= 6000:

                print("Fee Check: Passed")
                print("\nAdmission Status: SUCCESSFUL")
                print("Welcome to Middle School")

            elif fee >= 4000:

                print("Fee Check: Partially Passed")
                print("Please pay remaining fee")
                print("Admission Status: PENDING")

            else:

                print("Fee Check: Failed")
                print("Admission Status: NOT SUCCESSFUL")

        elif student_class >= 9 and student_class <= 10:

            print("Class Check: High School")

            if fee >= 7000:

                print("Fee Check: Passed")
                print("\nAdmission Status: SUCCESSFUL")
                print("Welcome to High School")

            elif fee >= 5000:

                print("Fee Check: Partially Passed")
                print("Please pay remaining fee")
                print("Admission Status: PENDING")

            else:

                print("Fee Check: Failed")
                print("Admission Status: NOT SUCCESSFUL")

        elif student_class == 11:

            print("Class Check: Senior Secondary")

            if fee >= 8000:

                print("Fee Check: Passed")
                print("\nAdmission Status: SUCCESSFUL")
                print("Welcome to Class 11")

            elif fee >= 6000:

                print("Fee Check: Partially Passed")
                print("Please pay remaining fee")
                print("Admission Status: PENDING")

            else:

                print("Fee Check: Failed")
                print("Admission Status: NOT SUCCESSFUL")

        elif student_class == 12:

            print("Class Check: Senior Secondary")

            if fee >= 8000:

                print("Fee Check: Passed")
                print("\nAdmission Status: SUCCESSFUL")
                print("Welcome to Class 12")

            elif fee >= 6000:

                print("Fee Check: Partially Passed")
                print("Please pay remaining fee")
                print("Admission Status: PENDING")

            else:

                print("Fee Check: Failed")
                print("Admission Status: NOT SUCCESSFUL")

        else:

            print("Invalid Class")
            print("Admission Status: NOT SUCCESSFUL")

    else:

        print("Age Check: Failed")
        print("Student age is too low")
        print("Minimum age required is 5 years")
        print("Admission Status: NOT SUCCESSFUL")


    print("\n==============================================")
    print("              FINAL INFORMATION")
    print("==============================================")

    print("Student Name :", name)
    print("Student Class:", student_class)
    print("Student Age  :", age)

    if age >= 5:

        if student_class >= 1 and student_class <= 5:

            print("School Section: Primary")

        elif student_class >= 6 and student_class <= 8:

            print("School Section: Middle")

        elif student_class >= 9 and student_class <= 10:

            print("School Section: High School")

        elif student_class >= 11 and student_class <= 12:

            print("School Section: Senior Secondary")

        else:

            print("School Section: Invalid")

    else:

        print("School Section: Not Eligible")


    print("\n==============================================")
    print("        THANK YOU FOR USING THE SYSTEM")
    print("==============================================")

if __name__ == '__main__':
    main()
```
